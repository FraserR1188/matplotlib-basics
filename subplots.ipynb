```python
# importing the libraries
import matplotlib
import matplotlib.pyplot as plt
import numpy as np
%matplotlib inline
```

```python
# create arrays with 5 random values between 0 and 50
a = np.random.randint(0, high=50, size=5)
b = np.random.randint(0, high=50, size=5)
x = np.random.randint(0, high=50, size=5)
y = np.random.randint(0, high=50, size=5)
print(a, b, x, y)
```

```python
# create a figure with one plot
fig = plt.figure()
ax = plt.subplot()
ax.plot(a, b, '.')
plt.show()
```
