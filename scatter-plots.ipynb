```python
%matplotlib widget

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
taxi_trips = pd.read_csv("Yellow Taxi Trips - Week 1.csv")
taxi_trips.head()
```

```python
#subsetting the data frame
JFK_trips = taxi_trips[(taxi_trips['Payment_Type'] == 'Credit card') & (taxi_trips['Rate_Code'] == 'JFK') & (taxi_trips['Passenger_Count'] > 0) & (taxi_trips['Total_Amount'] < 90)]
```

```python
# caluculating the descritive statistics
JFK_trips[['Tip_Amount', 'Total_Amount', 'Passenger_Count']].describe().round(2)
```

```python
# creating a scatter plot using the scripting layer
plt.figure(figsize=(9, 4))
plt.scatter(JFK_trips['Total_Amount'], JFK_trips['Tip_Amount'], marker = 'o', alpha = 0.5, s = 20, cmap = 'plasma', c = JFK_trips['Passenger_Count'])
plt.title('Scatter Plot of Tip Amount vs Total Amount')
plt.xlabel('Total Amount')
plt.ylabel('Tip Amount')
plt.colorbar(orientation = 'horizontal', label = 'Passenger Count')
plt.show()
```
