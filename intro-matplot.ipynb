```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
plt.plot(2, 2, '.')
plt.show()
```

```python
plt.xlim([0, 5])
plt.ylim([0, 5])
```

```python
x = np.array([1, 2, 3, 4, 5])
y = x*10
plt.plot(x, y, 'o-')
plt.show()
```

```python
plt.title('My First Plot', family ='serif', fontsize=14, color='blue', loc='left')
plt.xlabel('x-axis')
plt.ylabel('y-axis')
```
